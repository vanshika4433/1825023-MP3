```python
#rotation matrix

import matplotlib.pyplot as plt
import math as mt
import numpy as np

x = np.array([[np.cos(mt.radians(15)),np.sin(mt.radians(15))],
       [-np.sin(mt.radians(15)),np.cos(mt.radians(15))]])


a = np.array([9,9])
plt.scatter(a[0],a[1])
for i in range(24):
  f=x@a
  plt.scatter(f[0],f[1],color="purple")
  a=f
plt.show()
```

```python
# nth root of complex number

import numpy as np

z = 4 + 4j
n = 3

r = abs(z)
theta = np.angle(z)

for k in range(n):
    angle = (theta + 2*np.pi*k)/n

    root = r**(1/n) * (np.cos(angle) + 1j*np.sin(angle))

    print("Root =", root)
```

```python
# gauss lagendre
import numpy as np
import math

def f(x):
    return np.exp(x)/(1+x**2)

w = np.array([1, 1])

x = np.array([1/math.sqrt(3), -1/math.sqrt(3)])

f_x = f(x)

result = np.dot(w, f_x)

print("Result =", result)
```

```python
#gauss laguerre

import numpy as np

x, w = np.polynomial.laguerre.laggauss(3)

def f(x):
    return x**2

result = np.sum(w*f(x))

print("Result =", result)
```

```python
#fourier transform

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 100)

y = np.exp(-x**2)

Y = np.fft.fft(y)

plt.plot(x, y)
plt.title("Function exp(-x²)")
plt.grid()
plt.show()

plt.plot(np.abs(Y))
plt.title("FFT of exp(-x²)")
plt.xlabel("Frequency")
plt.ylabel("Amplitude")
plt.grid()
plt.show()
```

```python
#inverse Fourier transform

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 100)

# Original function
y = np.exp(-x**2)

# Fourier Transform
Y = np.fft.fft(y)

# Inverse Fourier Transform
y1 = np.fft.ifft(Y)

# Original graph
plt.plot(x, y)
plt.title("Original Function exp(-x²)")
plt.grid()
plt.show()

# Inverse Fourier Transform graph
plt.plot(x, y1.real)
plt.title("Inverse Fourier Transform")
plt.xlabel("x")
plt.ylabel("Amplitude")
plt.grid()
plt.show()
```

```python
#finite difference method

import numpy as np
import matplotlib.pyplot as plt

n = 10
h = 1/(n+1)

x = np.linspace(0, 1, n+2)

A = np.zeros((n,n))
B = np.zeros(n)

for i in range(n):
    A[i,i] = -2

    if i > 0:
        A[i,i-1] = 1

    if i < n-1:
        A[i,i+1] = 1

    B[i] = -2*h**2

y = np.zeros(n+2)

y[1:-1] = np.linalg.solve(A, B)

print("x =", x)
print("y =", y)

plt.plot(x, y, 'o-')
plt.title("Finite Difference Method")
plt.xlabel("x")
plt.ylabel("y")
plt.grid()
plt.show()
```

```python
#shooting method for linear bvp

import numpy as np
import matplotlib.pyplot as plt

def f(x, y, z):
    return z

def g(x, y, z):
    return -2

h = 0.1
x = 0
y = 0
z = 1

X = [x]
Y = [y]

while x < 1:

    k1 = h*f(x,y,z)
    l1 = h*g(x,y,z)

    k2 = h*f(x+h/2,y+k1/2,z+l1/2)
    l2 = h*g(x+h/2,y+k1/2,z+l1/2)

    k3 = h*f(x+h/2,y+k2/2,z+l2/2)
    l3 = h*g(x+h/2,y+k2/2,z+l2/2)

    k4 = h*f(x+h,y+k3,z+l3)
    l4 = h*g(x+h,y+k3,z+l3)

    y = y+(k1+2*k2+2*k3+k4)/6
    z = z+(l1+2*l2+2*l3+l4)/6

    x = round(x+h, 10)

    X.append(x)
    Y.append(y)

print("Final value of y =", y)

plt.plot(X,Y,'o-')
plt.xlabel("x")
plt.ylabel("y")
plt.title("Shooting Method using RK4")
plt.grid()
plt.show()
```

```python
#shooting method for non linear bvp

import numpy as np
import matplotlib.pyplot as plt

h = 0.1
x = np.arange(0, 1+h, h)

s = 0.5
y = 0
z = s
Y = [y]

for i in range(len(x)-1):

    k1 = h*z
    l1 = h*(-y**2)

    k2 = h*(z+l1/2)
    l2 = h*(-(y+k1/2)**2)

    k3 = h*(z+l2/2)
    l3 = h*(-(y+k2/2)**2)

    k4 = h*(z+l3)
    l4 = h*(-(y+k3)**2)

    y = y+(k1+2*k2+2*k3+k4)/6
    z = z+(l1+2*l2+2*l3+l4)/6

    Y.append(y)

print("Final value =", y)

plt.plot(x, Y, 'o-')
plt.xlabel("x")
plt.ylabel("y")
plt.title("Non-Linear BVP using RK4")
plt.grid()
plt.show()
```
